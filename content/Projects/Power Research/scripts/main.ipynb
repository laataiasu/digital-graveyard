```python
from neo4j import GraphDatabase

# Connect to Neo4j
uri = "neo4j://localhost:7687"
username = "neo4j"
password = "password"  # replace with your Neo4j password

driver = GraphDatabase.driver(uri, auth=(username, password))

cypher_script = """
// Persons
MERGE (peter:Person {name: "Peter Thiel"});
MERGE (naval:Person {name: "Naval Ravikant"});

// Companies & Funds
MERGE (paypal:Company {name: "PayPal"});
MERGE (palantir:Company {name: "Palantir"});
MERGE (foundersFund:Fund {name: "Founders Fund"});
MERGE (angellist:Company {name: "AngelList"});
MERGE (thielFellowship:Fund {name: "Thiel Fellowship"});

// Investments
MERGE (facebook:Company {name: "Meta (Facebook)"});
MERGE (twitter:Company {name: "Twitter"});
MERGE (yammer:Company {name: "Yammer"});

// Concepts
MERGE (angelInvesting:Concept {name: "Angel Investing"});
MERGE (ventureCapital:Concept {name: "Venture Capital"});
MERGE (philosophy:Concept {name: "Philosophy"});

// Relationships
MERGE (peter)-[:CO_FOUNDER_OF]->(paypal);
MERGE (peter)-[:FOUNDED]->(palantir);
MERGE (peter)-[:FOUNDED]->(foundersFund);
MERGE (peter)-[:FOUNDED]->(thielFellowship);
MERGE (peter)-[:INVESTED_IN]->(facebook);
MERGE (peter)-[:INVESTED_IN]->(yammer);
MERGE (peter)-[:ASSOCIATED_WITH]->(ventureCapital);
MERGE (peter)-[:PROMOTES]->(philosophy);

MERGE (naval)-[:FOUNDED]->(angellist);
MERGE (naval)-[:INVESTED_IN]->(twitter);
MERGE (naval)-[:INVESTED_IN]->(yammer);
MERGE (naval)-[:ASSOCIATED_WITH]->(angelInvesting);
MERGE (naval)-[:ASSOCIATED_WITH]->(ventureCapital);
MERGE (naval)-[:PROMOTES]->(philosophy);

// Cross-link
MERGE (peter)-[:KNOWS]->(naval);
"""

def seed_graph(tx):
    tx.run(cypher_script)

with driver.session() as session:
    session.write_transaction(seed_graph)

print("Graph seeded successfully.")
driver.close()
```
